```python

n = int(input("Enter number of machines: "))

machines = {}
Q = {}
N = {}

# Input rewards
for i in range(n):
    name = input(f"\nEnter machine {i+1} name: ")
    rewards = list(map(int, input("Enter rewards: ").split()))

    machines[name] = rewards

# 1. Arithmetic Mean Method
print("\n===== ARITHMETIC MEAN METHOD =====")

for name, rewards in machines.items():
    Q[name] = sum(rewards) / len(rewards)

    print(f"\nMachine {name}")
    print("Rewards:", rewards)
    print(f"Q({name}) = {Q[name]:.2f}")

selected = max(Q, key=Q.get)
print("\nGreedy Selected Machine:", selected)


# 2. Incremental Update Method
print("\n===== INCREMENTAL UPDATE METHOD =====")

Q = {}
N = {}

# Initialize values
for name in machines:
    Q[name] = 0.0
    N[name] = 0

# Update after every reward
for name, rewards in machines.items():
    print(f"\nMachine {name}")

    for R in rewards:
        N[name] += 1
        Q[name] = Q[name] + (R - Q[name]) / N[name]

        print(f"Reward = {R}, N = {N[name]}, Q = {Q[name]:.2f}")

    print(f"Final Q({name}) = {Q[name]:.2f}")

    selected = max(Q, key=Q.get)
    print("Current Greedy Selected Machine:", selected)


# Final Greedy Selection
print("\n===== FINAL RESULT =====")

for name in Q:
    print(f"Q({name}) = {Q[name]:.2f}")

selected = max(Q, key=Q.get)
print("\nGreedy Selected Machine:", selected)
```
