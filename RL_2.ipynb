```python

import random

# Part A: Machine Initialization
machines = ['A', 'B', 'C']
epsilon = 0.2
N_play = 10

# Part B: Initialize N, R and Q
N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': [], 'B': [], 'C': []}
Q = {'A': 0.0, 'B': 0.0, 'C': 0.0}

total_reward = 0

# Part C: Epsilon-Greedy Strategy
for play in range(1, N_play + 1):

    r = random.random()

    if r < epsilon:
        # Exploration
        selected = random.choice(machines)
        print("\nExploration")

    else:
        # Exploitation
        max_q = max(Q.values())
        best_machines = [m for m in machines if Q[m] == max_q]
        selected = random.choice(best_machines)
        print("\nExploitation")

    print("Play:", play)
    print("Random Number:", round(r, 2))
    print("Selected Machine:", selected)

    # Part D: Reward Collection
    reward = float(input("Enter reward: "))

    # Part E: Q-Value Update
    N[selected] += 1
    R[selected].append(reward)

    Q[selected] = Q[selected] + (reward - Q[selected]) / N[selected]

    total_reward += reward

    print("Updated N:", N)
    print("Updated Q:", {m: round(Q[m], 2) for m in machines})


# Part F: Display Final Results
print("\n========== FINAL RESULTS ==========")

print("\nMachine\tN\tRewards\t\tQ Value")

for m in machines:
    print(m, "\t", N[m], "\t", R[m], "\t", round(Q[m], 2))

print("\nTotal Reward Received:", total_reward)

print("\nBest Machine:", max(Q, key=Q.get))
```
